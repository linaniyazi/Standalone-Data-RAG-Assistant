### with SQL

```python
import sqlite3
conn = sqlite3.connect("superstore.db")
df.to_sql("orders", conn, if_exists="replace", index=False)
```

```python
with open("superstore_queries.sql", "r") as f:
    sql_script = f.read()

raw_queries = sql_script.split(";")

queries = []
for q in raw_queries:
    lines = q.strip().split("\n")
    clean_lines = [line for line in lines if line.strip() and not line.strip().startswith("--")]
    clean_query = " ".join(clean_lines).strip()
    if clean_query:
        queries.append(clean_query)

print(len(queries))
print(queries[0])
```

```python
for q in queries:
    print(q)
    result = pd.read_sql_query(q, conn)
    print(result)
    print("---")
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("superstore.db")

comparison_rows = []

# 1. Top category by total sales
pandas_result = df.groupby('Category')['Sales'].sum().idxmax()
sql_result = pd.read_sql_query(
    "SELECT Category FROM orders GROUP BY Category ORDER BY SUM(Sales) DESC LIMIT 1", conn
).iloc[0, 0]
comparison_rows.append(["1. Top category by sales", pandas_result, sql_result])

# 2. Average profit per region (comparing West as example)
pandas_result = round(df.groupby('Region')['Profit'].mean()['West'], 2)
sql_result = round(pd.read_sql_query(
    "SELECT AVG(Profit) FROM orders WHERE Region = 'West'", conn
).iloc[0, 0], 2)
comparison_rows.append(["2. Avg profit - West region", pandas_result, sql_result])

# 3. Sub-category with biggest loss
pandas_result = df.groupby('Sub-Category')['Profit'].sum().idxmin()
sql_result = pd.read_sql_query(
    'SELECT "Sub-Category" FROM orders GROUP BY "Sub-Category" ORDER BY SUM(Profit) ASC LIMIT 1', conn
).iloc[0, 0]
comparison_rows.append(["3. Sub-category biggest loss", pandas_result, sql_result])

# 4. Number of orders per segment (comparing Consumer as example)
pandas_result = df.groupby('Segment')['Order ID'].nunique()['Consumer']
sql_result = pd.read_sql_query(
    "SELECT COUNT(DISTINCT \"Order ID\") FROM orders WHERE Segment = 'Consumer'", conn
).iloc[0, 0]
comparison_rows.append(["4. Order count - Consumer segment", pandas_result, sql_result])

# 5. Highest sale value per region (comparing West as example)
pandas_result = df.groupby('Region')['Sales'].max()['West']
sql_result = pd.read_sql_query(
    "SELECT MAX(Sales) FROM orders WHERE Region = 'West'", conn
).iloc[0, 0]
comparison_rows.append(["5. Max sale - West region", pandas_result, sql_result])

# 6. Lowest sale value per region (comparing West as example)
pandas_result = df.groupby('Region')['Sales'].min()['West']
sql_result = pd.read_sql_query(
    "SELECT MIN(Sales) FROM orders WHERE Region = 'West'", conn
).iloc[0, 0]
comparison_rows.append(["6. Min sale - West region", pandas_result, sql_result])

# 7. Average sales per region (comparing West as example)
pandas_result = round(df.groupby('Region')['Sales'].mean()['West'], 2)
sql_result = round(pd.read_sql_query(
    "SELECT AVG(Sales) FROM orders WHERE Region = 'West'", conn
).iloc[0, 0], 2)
comparison_rows.append(["7. Avg sales - West region", pandas_result, sql_result])

# 8. Record count per city (comparing New York City as example)
pandas_result = df.groupby('City')['Order ID'].count()['New York City']
sql_result = pd.read_sql_query(
    "SELECT COUNT(*) FROM orders WHERE City = 'New York City'", conn
).iloc[0, 0]
comparison_rows.append(["8. Record count - New York City", pandas_result, sql_result])

# 9. Record count per country
pandas_result = df.groupby('Country')['Order ID'].count().iloc[0]
sql_result = pd.read_sql_query(
    "SELECT COUNT(*) FROM orders GROUP BY Country LIMIT 1", conn
).iloc[0, 0]
comparison_rows.append(["9. Record count - Country", pandas_result, sql_result])

# 10. Region with highest total profit
pandas_result = df.groupby('Region')['Profit'].sum().idxmax()
sql_result = pd.read_sql_query(
    "SELECT Region FROM orders GROUP BY Region ORDER BY SUM(Profit) DESC LIMIT 1", conn
).iloc[0, 0]
comparison_rows.append(["10. Region highest profit", pandas_result, sql_result])

comparison_df = pd.DataFrame(comparison_rows, columns=["Question", "Pandas Result", "SQL Result"])
comparison_df["Match"] = comparison_df["Pandas Result"] == comparison_df["SQL Result"]
comparison_df
```
