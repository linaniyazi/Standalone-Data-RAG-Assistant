```python
#make sure the equation Price Per Unit * Quantity = Total Spent is correct
check = (df_clean['Price Per Unit'] * df_clean['Quantity']).round(2)
mismatch = (check != df_clean['Total Spent'].round(2)).sum()
print(f"number of mismatched rows: {mismatch}")
```

```python
before_after = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Missing - Item",
        "Missing - Price Per Unit",
        "Missing - Quantity",
        "Missing - Total Spent",
        "Missing - Discount Applied",
        "Duplicate rows"
    ],
    "Before": [
        len(df),
        df['Item'].isnull().sum(),
        df['Price Per Unit'].isnull().sum(),
        df['Quantity'].isnull().sum(),
        df['Total Spent'].isnull().sum(),
        df['Discount Applied'].isnull().sum(),
        df.duplicated().sum()
    ],
    "After": [
        len(df_clean),
        df_clean['Item'].isnull().sum(),
        df_clean['Price Per Unit'].isnull().sum(),
        df_clean['Quantity'].isnull().sum(),
        df_clean['Total Spent'].isnull().sum(),
        df_clean['Discount Applied'].isnull().sum(),
        df_clean.duplicated().sum()
    ]
})
before_after
```
